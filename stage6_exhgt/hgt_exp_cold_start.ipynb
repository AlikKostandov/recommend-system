# HGT experiment: synthetic cold start

Ноутбук для cold-start экспериментов.

```
from pathlib import Path
import sys

sys.path.append(str(Path.cwd()))

from hgt_common import *

DATA_DIR = Path("..")  # поменяй на Path(".") или абсолютный путь при необходимости
CFG = HGTExperimentConfig(
    k_values=[20],
    target_metrics=["recall", "ndcg"],
    epochs=100,
)
set_seed(CFG.seed)
log_step(f"device: {CFG.device}")
log_step(f"target k_values: {CFG.k_values}")
log_step(f"target metrics: {CFG.target_metrics}")

tables = load_hgt_tables(DATA_DIR)
runner = make_hgt_runner(tables, CFG)
```

## Synthetic cold start

Оценка HGT в cold-start сценарии. Метрики: Recall@20 и NDCG@20.

```
STAGE = "genres_directors_actors_countries_tags_age_group_occupation"
THRESHOLD = 5.0
MIN_POS = 5

cold_start_results = runner.run_cold_start_grid(
    stage=STAGE,
    threshold=THRESHOLD,
    min_pos=MIN_POS,
)
cold_start_results.to_csv("hgt_cold_start_recall20_ndcg20_results.csv", index=False)
cold_start_results
```
